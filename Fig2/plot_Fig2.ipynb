# Figure 2: Spatial and running rainfall correlations

This notebook combines the spatial correlations for P1, P2, and P2 − P1 with
the 15-year running correlation for the Indonesia target box (95–125°E,
6°S–2°N).  It uses the two cached datasets in `data/`, so no raw precipitation
archives are required.


```python
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import cartopy.crs as ccrs
from cartopy.mpl.ticker import LongitudeFormatter, LatitudeFormatter

# ===== TEXT-SIZE CONTROLS =====
# Change the values below to resize all default text in the figure.
# Use axes.titlesize for panel titles; xtick/ytick.labelsize for map and plot ticks.
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Helvetica', 'Arial', 'DejaVu Sans'],
    'font.size': 24,
    'axes.labelsize': 24,
    'axes.titlesize': 24,
    'xtick.labelsize': 24,
    'ytick.labelsize': 24,
    'legend.fontsize': 24,
    'figure.dpi': 300,
})

# Run this notebook with Scripts/Fig2 as the working directory.
FIG2_DIR = Path('/Users/rizzie/Work/PaperENSO/Scripts_v2/Fig2')
DATA_DIR = FIG2_DIR if not (FIG2_DIR / 'data').exists() else (FIG2_DIR / 'data')
SPATIAL_CACHE = DATA_DIR / 'rainfall_corr_cache_1981_2025.nc'
RUNNING_CORR_CACHE = DATA_DIR / 'gambarL.2_running_corr_box1.csv'
OUTPUT_PATH = FIG2_DIR / 'Fig2_combined.png'

assert SPATIAL_CACHE.exists(), f'Missing {SPATIAL_CACHE}'
assert RUNNING_CORR_CACHE.exists(), f'Missing {RUNNING_CORR_CACHE}'
```

## Load the plotted data

`rainfall_corr_cache_1981_2025.nc` contains the P1 (1981–2006), P2
(2007–2025), and difference maps. The CSV stores the precomputed running
correlations used by the original `gambarL.2` notebook.


```python
rain_ds = xr.open_dataset(SPATIAL_CACHE)
running_corr_df = pd.read_csv(RUNNING_CORR_CACHE)

rain_corr_p1 = rain_ds['rain_corr_past']
rain_corr_p2 = rain_ds['rain_corr_recent']
rain_corr_diff = rain_ds['rain_corr_recent_minus_past']

rain_corr_p1_sig = rain_ds['rain_corr_past_sig']
rain_corr_p2_sig = rain_ds['rain_corr_recent_sig']
rain_corr_diff_sig = rain_ds['rain_corr_recent_minus_past_sig']

running_corr_df.groupby(['window', 'dataset']).size().unstack('dataset')
```

## Create the 2×2 composite

Panels (a)–(c) retain the spatial-correlation styling from `gambar4.2`; panel
(d) is the 15-year series from `gambarL.2`.


```python
corr_levels = np.arange(-1, 1.01, 0.05)
corr_ticks = np.arange(-1, 1.01, 0.25)
mc_extent = [80, 160, -20, 20]
mc_xticks = np.arange(90, 161, 20)
mc_yticks = np.arange(-20, 21, 10)
vimfc_domain = [95, 125, -6, 2]
dataset_order = ['MSWEP', 'ERA5', 'CHIRPS', 'GPCC', 'SAOBS', 'GPCP']
dataset_colors = {
    'MSWEP': '#1e3d59', 'ERA5': '#17b978', 'CHIRPS': '#ff5e62',
    'GPCC': '#ff9900', 'SAOBS': '#a06ee1', 'GPCP': '#085f56',
}

# ===== FIGURE LAYOUT CONTROLS =====
# figsize controls the overall width and height (inches).
# hspace increases/decreases the vertical space between the map rows.
# wspace increases/decreases the horizontal space between the left and right panels.
fig = plt.figure(figsize=(19, 15))
gs = fig.add_gridspec(2, 4, hspace=0.1, wspace=0.35)
ax_p1 = fig.add_subplot(gs[0, 0:2], projection=ccrs.PlateCarree(central_longitude=180))
ax_p2 = fig.add_subplot(gs[0, 2:4], projection=ccrs.PlateCarree(central_longitude=180))
ax_diff = fig.add_subplot(gs[1, 0:2], projection=ccrs.PlateCarree(central_longitude=180))
ax_running = fig.add_subplot(gs[1, 2:4])

map_panels = [
    (ax_p1, rain_corr_p1, rain_corr_p1_sig, '(a) Rainfall correlation: P1', 'bwr', 'neither'),
    (ax_p2, rain_corr_p2, rain_corr_p2_sig, '(b) Rainfall correlation: P2', 'bwr', 'neither'),
    (ax_diff, rain_corr_diff, rain_corr_diff_sig, '(c) Rainfall correlation difference: P2 − P1', 'BrBG', 'both'),
]

top_img = None
diff_img = None
for ax, data, sig_mask, title, cmap, extend in map_panels:
    img = data.reset_coords(drop=True).plot.pcolormesh(
        ax=ax, transform=ccrs.PlateCarree(), cmap=cmap, levels=corr_levels,
        extend=extend, add_colorbar=False, add_labels=False, infer_intervals=True,
    )
    if ax is ax_diff:
        diff_img = img
    else:
        top_img = img

    # Coarsening keeps the significance stippling legible at figure scale.
    sig_plot = (sig_mask.fillna(False).astype(np.int8)
                .coarsen(lat=8, lon=8, boundary='trim').max() > 0)
    yy, xx = np.where(sig_plot.values)
    ax.scatter(sig_plot['lon'].values[xx], sig_plot['lat'].values[yy], s=8,
               c='k', marker='.', linewidths=0, alpha=0.55,
               transform=ccrs.PlateCarree(), zorder=4)

    ax.coastlines(resolution='50m', linewidth=0.6, color='black', zorder=3)
    ax.set_extent(mc_extent, crs=ccrs.PlateCarree())
    if ax is ax_diff:
        ax.add_patch(Rectangle(
            (vimfc_domain[0], vimfc_domain[2]),
            vimfc_domain[1] - vimfc_domain[0],
            vimfc_domain[3] - vimfc_domain[2],
            fill=False, edgecolor='black', linewidth=2.8,
            transform=ccrs.PlateCarree(), zorder=5,
        ))
    ax.gridlines(draw_labels=False, linewidth=0.3, color='gray', alpha=0.4, linestyle='--')
    ax.set_xticks(mc_xticks, crs=ccrs.PlateCarree())
    ax.set_yticks(mc_yticks, crs=ccrs.PlateCarree())
    ax.xaxis.set_major_formatter(LongitudeFormatter(cardinal_labels={'east': 'E', 'west': 'W'}))
    ax.yaxis.set_major_formatter(LatitudeFormatter(cardinal_labels={'north': 'N', 'south': 'S'}))
    # TWEAK: labelsize changes map-coordinate labels; pad moves the title vertically.
    ax.tick_params(direction='out', top=True, right=True, labelsize=24)
    ax.set_title(title, loc='left', fontweight='bold', pad=12)

window = 15
for name in dataset_order:
    series = running_corr_df[
        (running_corr_df['window'] == window) & (running_corr_df['dataset'] == name)
    ]
    # TWEAK: change the two linewidth values to emphasize/de-emphasize the lines.
    ax_running.plot(series['djf_year'], series['running_corr'],
                    linewidth=3.5 if name == 'MSWEP' else 2.0,
                    color=dataset_colors[name], label=name)

ax_running.axhline(0.0, color='grey', linewidth=1.0, linestyle='--')
ax_running.set_ylim(-1.0, 1.0)
ax_running.set_yticks([-1.0, -0.5, 0.0, 0.5, 1.0])
ax_running.set_xlim(1981, 2025)
# TWEAK: 1985, 1990, ..., 2020 are the displayed running-correlation years.
ax_running.set_xticks(np.arange(1985, 2021, 5))
ax_running.grid(True, linestyle='--', linewidth=0.5, alpha=0.4)
# TWEAK: labelpad moves the Year label farther below the x-axis tick labels.
ax_running.set_xlabel('Year', labelpad=18)
# Place panel (d)'s y-axis ticks and label on the right side.
ax_running.set_ylabel(r'$\it{r}$', labelpad=1)
ax_running.yaxis.tick_right()
ax_running.yaxis.set_label_position('right')
# TWEAK: pad moves panel (d)'s title; labelsize controls its axis tick labels.
ax_running.set_title('d) Running Correlation 15 years', loc='left', pad=15)
ax_running.tick_params(labelsize=24)
ax_running.legend(ncol=3, frameon=True, loc='upper left', fontsize=18, facecolor='white',
                  edgecolor='grey', framealpha=0.95, columnspacing=1.0,
                  handlelength=1.5)

# TWEAK: these margins control blank space at the four edges of the figure.
fig.subplots_adjust(left=0.06, right=0.97, bottom=0.10, top=0.94)

# Match panel (d)'s physical width and height to panel (c)'s rendered GeoAxes box.
# TWEAK: edit this block only if panel (d) should use a different size.
fig.canvas.draw()  # Resolve Cartopy's map aspect before reading panel (c)'s position.
pos_diff = ax_diff.get_position()
pos_running = ax_running.get_position()
ax_running.set_position([pos_running.x0, pos_diff.y0, pos_diff.width, pos_diff.height])

# Shared P1/P2 colorbar placed in the space between the map rows.
pos_p1, pos_p2, pos_diff = ax_p1.get_position(), ax_p2.get_position(), ax_diff.get_position()
# TWEAK: [left, bottom, width, height] controls the P1/P2 colorbar position and size.
top_cax = fig.add_axes([0.25, pos_p1.y0 - 0.070, 0.50, 0.022])
top_cbar = fig.colorbar(top_img, cax=top_cax, orientation='horizontal', ticks=corr_ticks, extend='neither')
top_cbar.set_label(r'$\it{r}$', rotation=0, labelpad=14, fontsize=24)
top_cbar.ax.tick_params(labelsize=24)
top_cbar.ax.yaxis.set_label_position('right')

# Difference colorbar belongs directly beneath panel (c).
# TWEAK: change -0.067 to move this colorbar up/down; 0.022 changes its thickness.
diff_cax = fig.add_axes([pos_diff.x0, pos_diff.y0 - 0.067, pos_diff.width, 0.022])
diff_cbar = fig.colorbar(diff_img, cax=diff_cax, orientation='horizontal',
                         ticks=corr_ticks, extend='both')
diff_cbar.set_label(r'$\Delta r$', rotation=0, labelpad=18, fontsize=24)
diff_cbar.ax.tick_params(labelsize=24)
diff_cbar.ax.yaxis.set_label_position('right')

fig.savefig(OUTPUT_PATH, dpi=300, bbox_inches='tight')
print(f'Saved: {OUTPUT_PATH}')
plt.show()
rain_ds.close()
```
